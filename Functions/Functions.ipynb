# Python Functions: From Basic to Intermediate

A function is a block of reusable code that performs a specific task. Functions help us avoid repeating code and make programs easier to manage.

In this notebook, we will learn:
- what is a function
- how to define and call a function
- parameters and arguments
- return values
- default arguments
- keyword arguments
- scope
- recursion
- intermediate examples


## 1. What is a Function?

A function is like a small machine inside your program. You give it input, it does work, and it may return a result.

### Example
```python
def greet():
    print("Hello, welcome!")

greet()
```


````python
def greet():
    print("Hello, welcome!")

greet()
greet()
````

## 2. Function Parameters and Arguments

A parameter is a variable defined in the function. An argument is the value passed when calling the function.

```python
def greet_user(name):
    print("Hello,", name)

greet_user("Ali")
greet_user("Sara")
```


````python
def greet_user(name):
    print("Hello,", name)

greet_user("Ali")
greet_user("Sara")
````

## 3. Return Values

A function can return a result to the caller using the `return` statement.

```python
def add(a, b):
    return a + b

result = add(5, 3)
print(result)
```


````python
def add(a, b):
    return a + b

result = add(5, 3)
print(result)

print(add(10, 20))
````

## 4. Default Arguments

You can give function parameters default values. If the caller does not pass them, the defaults are used.

```python
def greet(name="friend"):
    print("Hello,", name)

greet()
greet("Ayesha")
```


````python
def greet(name="friend"):
    print("Hello,", name)

greet()
greet("Ayesha")
````

## 5. Keyword Arguments

When calling a function, you can pass arguments by their names.

```python
def student_info(name, age, city):
    print(name, age, city)

student_info(name="Ali", age=20, city="Karachi")
```


````python
def student_info(name, age, city):
    print(name, age, city)

student_info(name="Ali", age=20, city="Karachi")
student_info(city="Lahore", name="Sara", age=22)
````

## 6. Scope of Variables

Variables defined inside a function are local to that function. Variables outside the function are global.

```python
x = 10

def show():
    y = 20
    print(x, y)

show()
```


````python
x = 10

def show():
    y = 20
    print("Global x:", x)
    print("Local y:", y)

show()
print("Outside function x:", x)
````

## 7. Returning Multiple Values

A function can return more than one value using tuples.

```python
def calc(a, b):
    add = a + b
    multiply = a * b
    return add, multiply

sum_result, product_result = calc(3, 4)
print(sum_result, product_result)
```


````python
def calc(a, b):
    add = a + b
    multiply = a * b
    return add, multiply

sum_result, product_result = calc(3, 4)
print(sum_result, product_result)
````

## 8. Recursion

A recursive function is a function that calls itself. It is useful for problems like factorials and Fibonacci numbers.

```python
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```


````python
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
````

## 9. Intermediate Examples

### Example 1: Check even or odd
```python
def check_even_odd(num):
    if num % 2 == 0:
        return "Even"
    return "Odd"

print(check_even_odd(10))
```

### Example 2: Calculator
```python
def calculator(a, b, operation):
    if operation == "+":
        return a + b
    elif operation == "-":
        return a - b
    elif operation == "*":
        return a * b
    elif operation == "/":
        return a / b
    else:
        return "Invalid operation"
```


````python
def check_even_odd(num):
    if num % 2 == 0:
        return "Even"
    return "Odd"

print(check_even_odd(10))
print(check_even_odd(7))

def calculator(a, b, operation):
    if operation == "+":
        return a + b
    elif operation == "-":
        return a - b
    elif operation == "*":
        return a * b
    elif operation == "/":
        return a / b
    else:
        return "Invalid operation"

print(calculator(10, 5, "+"))
print(calculator(10, 5, "/"))
````

- call it using its name
- pass data through parameters
- return results using `return`
- use default and keyword arguments
- understand function scope
- recursion can solve some problems elegantly

Functions make programs cleaner, shorter, and easier to maintain.
