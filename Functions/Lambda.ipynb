# Python Lambda Functions: Beginner to Intermediate

A lambda function is a small anonymous function in Python. It is defined using the keyword `lambda` instead of `def`.

Lambda functions are useful when you need a short function for a small task, especially with built-in functions like `map()`, `filter()`, and `sorted()`.

In this notebook, we will learn:
- what a lambda function is
- basic syntax
- using lambda with `map()`
- using lambda with `filter()`
- sorting with `key=lambda`
- intermediate examples


## 1. What is a Lambda Function?

A lambda function is a function without a name. It is usually used for short operations.

### Example
```python
square = lambda x: x * x
print(square(5))
```

This gives the same result as:
```python
def square(x):
    return x * x
```


````python
square = lambda x: x * x
print(square(5))

add = lambda a, b: a + b
print(add(3, 4))
````

## 2. Lambda Syntax

The syntax is:

```python
lambda arguments: expression
```

Examples:
```python
double = lambda x: x * 2
greet = lambda name: "Hello " + name
```


````python
double = lambda x: x * 2
print(double(6))

greet = lambda name: "Hello " + name
print(greet("Ali"))
````

## 3. Using Lambda with `map()`

The `map()` function applies a function to every item in a list.

```python
numbers = [1, 2, 3, 4]
squared = list(map(lambda x: x * x, numbers))
print(squared)
```


````python
numbers = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x * x, numbers))
print("Original:", numbers)
print("Squared:", squared)
````

## 4. Using Lambda with `filter()`

The `filter()` function keeps only the items that satisfy a condition.

```python
numbers = [1, 2, 3, 4, 5, 6]
evens = list(filter(lambda x: x % 2 == 0, numbers))
print(evens)
```


````python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
evens = list(filter(lambda x: x % 2 == 0, numbers))
print(evens)
````

## 5. Sorting with `key=lambda`

Lambda functions are often used in `sorted()` to sort items by custom rules.

```python
students = ["Ali", "Zain", "Ayesha"]
print(sorted(students))
print(sorted(students, key=lambda s: len(s)))
```


````python
students = ["Ali", "Zain", "Ayesha", "Saad"]
print(sorted(students))
print(sorted(students, key=lambda s: len(s)))
````

## 6. Intermediate Examples

### Example 1: Find highest number
```python
numbers = [10, 25, 7, 35, 18]
max_num = max(numbers, key=lambda x: x)
print(max_num)
```

### Example 2: Multiply all numbers in a list
```python
from functools import reduce
numbers = [1, 2, 3, 4]
product = reduce(lambda x, y: x * y, numbers)
print(product)
```

### Example 3: Sort dictionary by value
```python
scores = {"Ali": 90, "Sara": 85, "Zain": 95}
sorted_scores = sorted(scores.items(), key=lambda item: item[1])
print(sorted_scores)
```


````python
# Example 1: Find highest number
numbers = [10, 25, 7, 35, 18]
max_num = max(numbers, key=lambda x: x)
print(max_num)

print("---")

# Example 2: Multiply all numbers in a list
from functools import reduce
values = [1, 2, 3, 4]
product = reduce(lambda x, y: x * y, values)
print(product)

print("---")

# Example 3: Sort dictionary by value
scores = {"Ali": 90, "Sara": 85, "Zain": 95}
sorted_scores = sorted(scores.items(), key=lambda item: item[1])
print(sorted_scores)
````

## Final Summary

Lambda functions are simple, one-line functions used for short operations. They are especially helpful with `map()`, `filter()`, `sorted()`, and other functional programming tools.

Key points:
- lambda functions are anonymous
- syntax: `lambda arguments: expression`
- useful for short and quick operations
- commonly used with `map()`, `filter()`, and `sorted()`
- great for simple logic without writing a full function

Use lambda functions when the logic is small and straightforward.
