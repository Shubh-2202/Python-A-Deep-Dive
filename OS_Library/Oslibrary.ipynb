# Python `os` Module: Beginner to Intermediate

Python's built-in `os` module lets your program interact with the operating system. You can inspect folders, work with paths, create directories, and read environment variables.

These examples use cross-platform functions where possible. Some results, such as folder contents and paths, depend on your computer.

## 1. Importing the `os` Module

Import `os` before using its functions. You can then call a function using `os.function_name()`.

```python
import os

print("Operating system:", os.name)
print("Current folder:", os.getcwd())
```

## 2. Current Directory and Folder Contents

- `os.getcwd()` returns the current working directory.
- `os.listdir(path)` returns the names of items in a directory.

If no path is given to `listdir()`, it lists the current working directory. The current working directory is not always the same as the folder where the notebook file is saved.

```python
current_folder = os.getcwd()
print("Current folder:", current_folder)

items = os.listdir(current_folder)
print("Items in the current folder:")
for item in items:
    print(item)
```

## 3. Building and Checking Paths

Use `os.path.join()` to build paths. It inserts the correct separator for the current operating system, so you do not need to manually write `/` or `\\`.

Useful path functions include `exists()`, `isfile()`, `isdir()`, `basename()`, and `dirname()`.

```python
folder = os.getcwd()
example_path = os.path.join(folder, "notes.txt")

print("Path:", example_path)
print("Exists:", os.path.exists(example_path))
print("Is a file:", os.path.isfile(example_path))
print("Is a folder:", os.path.isdir(folder))
print("File name:", os.path.basename(example_path))
print("Parent folder:", os.path.dirname(example_path))
```

## 4. Creating Directories

`os.mkdir(name)` creates one directory. `os.makedirs(path, exist_ok=True)` can create a directory and any missing parent directories. With `exist_ok=True`, it does not raise an error if the directory already exists.

This example creates a folder named `os_practice` inside the current working directory.

```python
practice_folder = os.path.join(os.getcwd(), "os_practice")
os.makedirs(practice_folder, exist_ok=True)
print("Practice folder is ready:", practice_folder)
```

## 5. Renaming and Removing Files or Folders

- `os.rename(old_path, new_path)` renames or moves a file or directory.
- `os.remove(path)` removes a file.
- `os.rmdir(path)` removes an empty directory.

These operations change or delete data. Check the path carefully before using them, and do not use `os.remove()` on a directory. The example below only renames a practice file that it creates itself.

```python
old_name = os.path.join(practice_folder, "draft.txt")
new_name = os.path.join(practice_folder, "final.txt")

with open(old_name, "w", encoding="utf-8") as file:
    file.write("A practice file for learning os.rename().\n")

os.rename(old_name, new_name)
print("Renamed file exists:", os.path.isfile(new_name))
```

## 6. Exploring Folders with `os.walk()`

`os.walk(folder)` visits a folder and its subfolders. Each step provides the current folder path, a list of subfolder names, and a list of file names. This is useful for searching or summarizing directory trees.

```python
for folder_path, folder_names, file_names in os.walk(practice_folder):
    print("Folder:", folder_path)
    print("Subfolders:", folder_names)
    print("Files:", file_names)
```

## 7. Environment Variables

Environment variables are settings provided by the operating system or the program that started Python. Read them with `os.environ` or `os.getenv()`.

Do not print or share environment variables indiscriminately: some may contain private information such as tokens or passwords. The example reads the commonly available `PATH` setting only to show how lookup works.

```python
path_setting = os.getenv("PATH")
print("PATH is available:", path_setting is not None)
```

## 8. Handling Operating-System Errors

Operations can fail, for example when a path does not exist or permission is denied. Python reports these problems as exceptions, often subclasses of `OSError`. Catch a specific error when you can handle it meaningfully.

```python
unknown_folder = os.path.join(os.getcwd(), "folder_that_does_not_exist")

try:
    print(os.listdir(unknown_folder))
except FileNotFoundError as error:
    print("Folder was not found:", error.filename)
```

## 9. `os` and `pathlib`

The `os` module provides many operating-system functions. For everyday path building and file or directory operations, Python's `pathlib` module often offers a readable object-oriented alternative.

For example, `Path.cwd()` is similar to `os.getcwd()`, and `Path / "file.txt"` builds a path. Both tools are useful; choose the one that makes your code clearest.

```python
from pathlib import Path

pathlib_folder = Path.cwd()
pathlib_file = pathlib_folder / "notes.txt"

print("Current folder:", pathlib_folder)
print("Example path:", pathlib_file)
print("Exists:", pathlib_file.exists())
```

## Final Summary

- Import the module with `import os`.
- Use `os.getcwd()` and `os.listdir()` to inspect folders.
- Use `os.path.join()` to build paths for different operating systems.
- Check paths with `exists()`, `isfile()`, and `isdir()`.
- Create folders with `os.makedirs()`.
- Use `os.rename()`, `os.remove()`, and `os.rmdir()` carefully because they change or delete data.
- Explore directory trees with `os.walk()`.
- Use `os.getenv()` to read a named environment variable, and protect sensitive values.
- Consider `pathlib` for readable path operations.
