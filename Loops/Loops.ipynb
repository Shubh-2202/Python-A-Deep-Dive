# Python Loops: For Loops and While Loops

Loops are used to repeat instructions multiple times in Python. They are very useful when you want to work with lists, numbers, strings, or repeated actions.

In this notebook, we will learn:
- for loops
- while loops
- break, continue, and pass
- loop with ranges
- nested loops
- common beginner and intermediate examples


## 1. For Loop

A `for` loop is used when you already know how many times you want to repeat something or when you want to go through items in a collection.

### Basic example
```python
for i in range(5):
    print(i)
```

This prints numbers from 0 to 4.


````python
for i in range(5):
    print(i)

print("---")

for fruit in ["apple", "banana", "mango"]:
    print(fruit)
````

## 2. For Loop with Lists and Strings

```python
names = ["Ali", "Sara", "Usman"]
for name in names:
    print(name)

for ch in "Python":
    print(ch)
```


````python
names = ["Ali", "Sara", "Usman"]
for name in names:
    print(name)

print("---")

for ch in "Python":
    print(ch)
````

## 3. Using `range()`

The `range()` function creates a sequence of numbers.

```python
for x in range(1, 10, 2):
    print(x)
```

This prints odd numbers from 1 to 9.


````python
for x in range(1, 10, 2):
    print(x)

print("---")

for x in range(10, 0, -1):
    print(x)
````

## 4. While Loop

A `while` loop runs as long as a condition is true.

### Basic example
```python
count = 1
while count <= 5:
    print(count)
    count += 1
```


````python
count = 1
while count <= 5:
    print(count)
    count += 1
````

## 5. Break, Continue, and Pass

### `break`
Stops the loop completely.

```python
for i in range(1, 10):
    if i == 5:
        break
    print(i)
```

### `continue`
Skips the current iteration and continues with the next one.

```python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

### `pass`
Does nothing; it is often used as a placeholder.


````python
print("Break example")
for i in range(1, 10):
    if i == 5:
        break
    print(i)

print("---")
print("Continue example")
for i in range(1, 6):
    if i == 3:
        continue
    print(i)

print("---")
for i in range(1, 4):
    pass
print("Pass used as placeholder")
````

## 6. For Loop vs While Loop

Use a `for` loop when:
- you know the number of iterations
- you want to iterate through a list, tuple, or string

Use a `while` loop when:
- the number of iterations is not fixed
- the loop depends on a condition

Example:
```python
# For loop: fixed count
for i in range(3):
    print("Hello")

# While loop: condition based
count = 0
while count < 3:
    print("Hello")
    count += 1
```


## 7. Intermediate Examples

### Example 1: Sum of first 10 numbers
```python
total = 0
for i in range(1, 11):
    total += i
print(total)
```

### Example 2: Find even numbers
```python
for i in range(1, 11):
    if i % 2 == 0:
        print(i)
```

### Example 3: While loop for password attempt
```python
attempts = 0
password = "1234"
user_input = ""

while user_input != password and attempts < 3:
    user_input = input("Enter password: ")
    attempts += 1
```


````python
# Example 1: Sum of first 10 numbers
total = 0
for i in range(1, 11):
    total += i
print("Sum:", total)

print("---")

# Example 2: Find even numbers
for i in range(1, 11):
    if i % 2 == 0:
        print(i, end=" ")
print()

print("---")

# Example 3: While loop with condition
count = 0
while count < 5:
    print("Count is", count)
    count += 1
````

## 8. Nested Loops

A nested loop is a loop inside another loop.

```python
for i in range(3):
    for j in range(2):
        print(i, j)
```


````python
for i in range(3):
    for j in range(2):
        print(f"i = {i}, j = {j}")
````

## Final Summary

Both `for` loops and `while` loops are used to repeat code.

- Use `for` loops for known collections or fixed counts.
- Use `while` loops for condition-based repetition.
- Use `break` to stop the loop, `continue` to skip, and `pass` as a placeholder.
- Nested loops are useful for working with tables and patterns.

Loops are one of the most important concepts in Python programming.
