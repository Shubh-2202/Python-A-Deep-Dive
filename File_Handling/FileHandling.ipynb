# Python File Handling: Beginner to Intermediate

File handling lets a Python program save information to files and read it later. In this notebook, we will practice working with text files safely, then look at paths, errors, CSV, and JSON.

The examples create files in the notebook's current working folder. You can check that folder with `Path.cwd()`.

## 1. Opening a File

Python uses `open()` to work with a file. The first argument is the file name; the second is the mode.

Common modes:
- `"r"`: read a file that already exists
- `"w"`: write a file; creates it or replaces its existing contents
- `"a"`: append to the end; creates the file if it does not exist
- `"x"`: create a new file, and raise an error if it already exists

Be careful with `"w"`: opening an existing file in this mode erases its old contents.

```python
# Write some text to a practice file.
# The file is created in the current working folder.
with open("practice.txt", "w", encoding="utf-8") as file:
    file.write("Hello, file!\n")
    file.write("This is my first file-handling example.\n")
```

## 2. Reading a File

After creating `practice.txt`, open it in read mode. `read()` returns all the text as one string. Use `print()` to display it.

```python
with open("practice.txt", "r", encoding="utf-8") as file:
    contents = file.read()

print(contents)
```

## 3. Why Use `with`?

The `with` statement is called a context manager. It closes the file automatically when the indented block ends, even if an error occurs. This is the recommended way to open files.

You may also read one line at a time with `readline()`, or loop over the file. Looping is useful for large files because it does not load the entire file into memory at once.

```python
# Read all lines into a list.
with open("practice.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()

print(lines)

# Loop over a file one line at a time.
with open("practice.txt", "r", encoding="utf-8") as file:
    for line in file:
        print(line.strip())  # strip() removes the newline around the text
```

## 4. Appending to a File

Append mode (`"a"`) adds new text at the end and keeps the existing contents. The newline `\n` moves the next text to a new line.

```python
with open("practice.txt", "a", encoding="utf-8") as file:
    file.write("This line was added later.\n")

with open("practice.txt", "r", encoding="utf-8") as file:
    print(file.read())
```

## 5. Working with File Paths

A relative path such as `"practice.txt"` is interpreted from the program's current working folder. The `pathlib` module provides a convenient, cross-platform way to build and inspect paths.

A path can refer to a file in another folder too. Make sure the parent folder exists before writing there.

```python
from pathlib import Path

file_path = Path("practice.txt")
print("Current folder:", Path.cwd())
print("File exists:", file_path.exists())
print("File name:", file_path.name)
print("File size:", file_path.stat().st_size, "bytes")
```

## 6. Handling File Errors

Trying to read a file that does not exist raises `FileNotFoundError`. Handle a specific error when you have a useful response for it; do not hide errors that you do not understand.

```python
missing_path = Path("missing_file.txt")

try:
    with missing_path.open("r", encoding="utf-8") as file:
        print(file.read())
except FileNotFoundError:
    print(f"The file was not found: {missing_path}")
```

## 7. Intermediate: Working with CSV Files

CSV files store table-shaped data as rows and columns. Python's built-in `csv` module handles quoting and separators more reliably than manually splitting each line.

```python
import csv

rows = [
    {"name": "Ali", "score": 88},
    {"name": "Sara", "score": 95},
]

with open("scores.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=["name", "score"])
    writer.writeheader()
    writer.writerows(rows)

with open("scores.csv", "r", newline="", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    for row in reader:
        print(row)
```

## 8. Intermediate: Working with JSON Files

JSON is a common format for storing and exchanging structured data. Python's `json` module can convert compatible dictionaries and lists to JSON text and load them back.

```python
import json

profile = {
    "name": "Ayesha",
    "age": 21,
    "skills": ["Python", "Problem solving"],
}

with open("profile.json", "w", encoding="utf-8") as file:
    json.dump(profile, file, indent=4)

with open("profile.json", "r", encoding="utf-8") as file:
    loaded_profile = json.load(file)

print(loaded_profile["name"])
print(loaded_profile["skills"])
```

## Final Summary

- Use `open()` to work with files and choose the correct mode.
- Prefer `with open(...)` so files close automatically.
- Use `"r"` to read, `"w"` to replace contents, and `"a"` to append.
- Specify `encoding="utf-8"` for text files.
- Use `pathlib.Path` to work with file paths.
- Handle expected errors such as `FileNotFoundError` explicitly.
- Use `csv` for table data and `json` for structured data.
