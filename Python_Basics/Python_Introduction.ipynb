

# Python Introduction

Python is a simple, powerful, and beginner-friendly programming language.
It is used for:
- Web development
- Data science
- Artificial intelligence
- Automation
- Game development
- Scripting

Python is easy to read because it uses simple English-like syntax.

## 1. Python Syntax
Syntax means the rules for writing correct Python code.

Example:
```python
print("Hello, World!")
```

Output:
```python
Hello, World!
```

Important syntax rules:
- Python is case-sensitive
- Each statement is usually written on a new line
- Strings must be inside quotes
- Use `=` to assign values
- Parentheses are used with functions

Example:
```python
name = "Alice"
age = 20
print(name)
print(age)
```

## 2. Variables
A variable is a container that stores data.

```python
name = "John"
age = 18
marks = 90
```

Variable names should:
- Start with a letter or underscore
- Not contain spaces
- Not use Python keywords

Example:
```python
student_name = "Sam"
total_marks = 500
```

## 3. Data Types
Python has different data types:

```python
name = "Python"      # string
age = 18             # integer
price = 25.5         # float
is_student = True    # boolean
```

Common data types:
- int → whole numbers
- float → decimal numbers
- str → text
- bool → True or False

## 4. Comments
Comments are notes in the code that Python ignores.

```python
# This is a comment
print("Hello")
```

## 5. User Input
Python can get input from the user using `input()`.

```python
name = input("Enter your name: ")
print("Hello", name)
```

Example output:
```python
Enter your name: Ali
Hello Ali
```

Important:
- `input()` always returns a string
- If you want a number, convert it

Example:
```python
age = int(input("Enter your age: "))
print("Your age is:", age)
```

Correct version of your code:
```python
age = int(input("Enter Your Age: "))
print("Your age is:", age)
```

Incorrect version:
```python
age=int(input("Enter Your Age:))
```

Why it is wrong:
- missing closing quote
- missing closing parenthesis
- spacing is not standard

## 6. Printing Output
Use `print()` to display output.

## 7. Arithmetic Operators
Python supports arithmetic operations:

```python
a = 10
b = 5

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a % b)
```

Output:
```python
15
5
50
2.0
0
```

## 8. Comparison Operators
Comparison operators compare values:

```python
x = 10
y = 5

print(x > y)
print(x == y)
print(x < y)
```

## 9. If, Else, Elif
Python uses conditions to make decisions.

```python
age = 18

if age >= 18:
    print("You are an adult.")
else:
    print("You are not an adult.")
```

Output:
```python
You are an adult.
```

Example with `elif`:
```python
marks = 85

if marks >= 90:
    print("Grade A")
elif marks >= 80:
    print("Grade B")
else:
    print("Grade C")
```

## 10. Indentation in Python
Indentation is very important in Python.

Python uses indentation to show blocks of code.

Correct example:
```python
if 5 > 3:
    print("5 is greater than 3")
```

Wrong example:
```python
if 5 > 3:
print("5 is greater than 3")
```

This will cause an error because the print statement is not indented.

Common indentation rule:
- Use 4 spaces
- Blocks of code belong under statements like `if`, `for`, `while`, and `else`

Example:
```python
age = 20

if age >= 18:
    print("Eligible")
    print("You can vote")
```

Output:
```python
Eligible
You can vote
```

## 11. Loops
Loops repeat code multiple times.

### For loop
```python
for i in range(5):
    print(i)
```

Output:
```python
0
1
2
3
4
```

### While loop
```python
count = 1

while count <= 3:
    print(count)
    count += 1
```

Output:
```python
1
2
3
```

## 12. Lists
A list stores multiple values.

```python
numbers = [10, 20, 30, 40]
print(numbers)
print(numbers[0])
```

Output:
```python
[10, 20, 30, 40]
10
```

## 15. Common Beginner Mistakes
### Mistake 1: Missing quotes
```python
print(Hello)
```
This is wrong.

Correct:
```python
print("Hello")
```

### Mistake 2: Missing closing parenthesis or quote
```python
age = int(input("Enter your age: ))
```

Correct:
```python
age = int(input("Enter your age: "))
```

### Mistake 3: Wrong indentation
```python
if 10 > 5:
print("Yes")
```

Correct:
```python
if 10 > 5:
    print("Yes")
```

### Mistake 4: Invalid variable name
```python
2name = "Alice"
```

Correct:
```python
name2 = "Alice"
```
