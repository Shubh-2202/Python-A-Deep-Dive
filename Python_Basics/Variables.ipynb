

A variable is a name that stores a value in memory.

Example:
```python
name = "Ali"
age = 20
```

Here:
- `name` is a variable
- `"Ali"` is the value stored in `name`
- `age` is a variable
- `20` is the value stored in `age`

Variables help us store and reuse data in programs.

## Why do we use variables?
Variables are used to:
- store data
- perform calculations
- store user input
- make programs dynamic

Example:
```python
x = 10
y = 5
sum_value = x + y
print(sum_value)
```

Output:
```python
15
```

## Variable assignment
In Python, we assign a value using `=`.

```python
number = 50
message = "Welcome"
```

## Variable naming rules
Variable names must follow some rules:

1. Can contain letters, numbers, and underscores
2. Cannot start with a number
3. Cannot contain spaces
4. Cannot use Python keywords like `if`, `for`, `class`
5. Variable names are case-sensitive

Examples:
```python
name = "Sara"        # valid
age = 18             # valid
student_name = "Ali" # valid
2name = "John"       # invalid
```

Correct examples:
```python
name1 = "John"
student_name = "Mary"
_age = 25
```

## Good variable naming style
Use meaningful names:

```python
student_name = "Aisha"
total_marks = 450
```

Avoid unclear names:
```python
x = "Aisha"
a = 450
```

Better practice is to use descriptive names.

## Variable types
Python variables can store different types of values.

```python
name = "Python"   # string
age = 18           # integer
price = 25.5       # float
is_student = True  # boolean
```

### String
```python
name = "Alice"
print(name)
```

### Integer
```python
age = 20
print(age)
```

### Float
```python
salary = 45000.50
print(salary)
```

### Boolean
```python
is_active = True
print(is_active)
```

## Reassigning variables
You can change the value of a variable later.

```python
x = 10
print(x)

x = 20
print(x)
```

Output:
```python
10
20
```

## Multiple assignment
You can assign values to multiple variables in one line.

```python
a, b, c = 10, 20, 30
print(a)
print(b)
print(c)
```

Output:
```python
10
20
30
```

You can also assign the same value to multiple variables:

```python
x = y = z = 5
print(x)
print(y)
print(z)
```

Output:
```python
5
5
5
```

## Variables and user input
Variables are often used to store input from the user.

```python
name = input("Enter your name: ")
print("Hello", name)
```

Example:
```python
Enter your name: Aisha
Hello Aisha
```

When using numbers from input, convert them to int or float:

```python
age = int(input("Enter your age: "))
print("Your age is:", age)
```

## Example program using variables
```python
name = input("Enter your name: ")
age = int(input("Enter your age: "))
city = input("Enter your city: ")

print("Name:", name)
print("Age:", age)
print("City:", city)
```

## Variable operations
Variables can be used in calculations.

```python
a = 10
b = 5

sum_value = a + b
diff = a - b
product = a * b
division = a / b

print("Sum:", sum_value)
print("Difference:", diff)
print("Product:", product)
print("Division:", division)
```

Output:
```python
Sum: 15
Difference: 5
Product: 50
Division: 2.0
```

## Local variables and global variables
A local variable is created inside a function.
A global variable is created outside a function.

Example:
```python
x = 10  # global variable

def show():
    y = 20  # local variable
    print(x)
    print(y)

show()
```

## Constants
Constants are variables whose values do not change.

In Python, we usually write constants in uppercase:

```python
PI = 3.14
GRAVITY = 9.8

print(PI)
print(GRAVITY)
```

## Variables in conditions
Variables are commonly used in `if` statements.

```python
age = 18

if age >= 18:
    print("You are an adult.")
else:
    print("You are not an adult.")
```

Output:
```python
You are an adult.
```

## Type conversion with variables
You can convert one variable type into another.

```python
num = "25"
print(type(num))

num = int(num)
print(type(num))
print(num + 5)
```

Output:
```python
<class 'str'>
<class 'int'>
30
```

## Common mistakes with variables
### 1. Using spaces in variable names
```python
student name = "Ali"
```
This is invalid because variable names cannot contain spaces.

Correct:
```python
student_name = "Ali"
```

### 2. Starting with a number
```python
2name = "Ali"
```
This is invalid.

Correct:
```python
name2 = "Ali"
```

### 3. Using Python keywords
```python
for = 10
```
This is invalid because `for` is a reserved keyword.

### 4. Forgetting to assign a value first
```python
print(age)
age = 20
```
This gives an error because the variable is used before assignment.
