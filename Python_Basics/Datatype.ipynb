
# Data Types in Python

Data types are the kinds of values that a variable can store.

Python has several built-in data types, including:
- Integer
- Float
- String
- Boolean
- List
- Tuple
- Set
- Dictionary

In this lesson, we will focus on the main basic data types.

## 1. Integer
An integer is a whole number, positive or negative.

```python
age = 18
marks = 90
```

Example:
```python
print(type(age))
```

Output:
```python
<class 'int'>
```

## 2. Float
A float is a number with decimal points.

```python
price = 25.5
height = 5.8
```

Example:
```python
print(type(price))
```

Output:
```python
<class 'float'>
```

## 3. String
A string is a sequence of characters enclosed in single or double quotes.

```python
name = "Alice"
city = 'Karachi'
```

Example:
```python
print(type(name))
```

Output:
```python
<class 'str'>
```

String example:
```python
print("Hello, World!")
```

Output:
```python
Hello, World!
```

## 4. Boolean
A boolean represents True or False.

```python
is_pass = True
is_student = False
```

Example:
```python
print(type(is_pass))
```

Output:
```python
<class 'bool'>
```

## 5. Type Checking
We can check the data type of a variable using `type()`.

```python
x = 10
print(type(x))
```

Output:
```python
<class 'int'>
```

Example with different values:
```python
a = 5
b = 5.7
c = "Python"
d = True

print(type(a))
print(type(b))
print(type(c))
print(type(d))
```

Output:
```python
<class 'int'>
<class 'float'>
<class 'str'>
<class 'bool'>
```

## 6. Type Conversion
Python allows converting one data type into another.

### Converting int to float
```python
num = 10
print(float(num))
```

Output:
```python
10.0
```

### Converting float to int
```python
price = 25.9
print(int(price))
```

Output:
```python
25
```

### Converting string to int
```python
age = "18"
print(int(age))
```

Output:
```python
18
```

### Converting int to string
```python
age = 18
print(str(age))
```

Output:
```python
"18"
```

## 7. Why Data Types Matter
Data types tell Python how to handle values.

Example:
```python
a = 10
b = 5
print(a + b)
```

Output:
```python
15
```

But if we do:
```python
a = "10"
b = "5"
print(a + b)
```

Output:
```python
105
```

This happens because strings are joined together, not added numerically.

So we must be careful about the data type.

## 8. Example program

Output:
```python
Name: Aisha
Age: 16
Fee: 2500.5
Is Student: True
```

## 9. Data types in user input
`input()` always returns a string by default.

```python
age = input("Enter your age: ")
print(type(age))
```

Output:
```python
<class 'str'>
```

Because `input()` returns text, we must convert it when needed.

Example:
```python
age = int(input("Enter your age: "))
print(type(age))
```

Output:
```python
<class 'int'>
```

## 10. Common beginner mistakes
### Mistake 1: Adding string and number
```python
a = "10"
b = 5
print(a + b)
```

This causes an error because string and int cannot be added directly.

Correct:
```python
a = int("10")
b = 5
print(a + b)
```

Output:
```python
15
```

### Mistake 2: Forgetting to convert input
```python
age = input("Enter age: ")
print(age + 1)
```

This creates an error because `age` is a string.

Correct:
```python
age = int(input("Enter age: "))
print(age + 1)
```

## 11. List Data Type
A list stores multiple values in one variable.

```python
numbers = [1, 2, 3, 4, 5]
print(numbers)
```

Output:
```python
[1, 2, 3, 4, 5]
```

## 12. Tuple Data Type
A tuple is similar to a list, but it is immutable (cannot be changed).

```python
point = (10, 20)
print(point)
```

Output:
```python
(10, 20)
```

## 13. Set Data Type
A set is an unordered collection of unique values.

```python
numbers = {1, 2, 3, 3, 4}
print(numbers)
```

Output:
```python
{1, 2, 3, 4}
```

## 14. Dictionary Data Type
A dictionary stores data in key-value pairs.

```python
student = {"name": "Ali", "age": 18}
print(student)
```

Output:
```python
{'name': 'Ali', 'age': 18}
```
