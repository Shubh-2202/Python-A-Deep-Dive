
# Operators in Python

Operators are symbols used to perform operations on values or variables.

Python supports different types of operators, such as:
- Arithmetic operators
- Comparison operators
- Assignment operators
- Logical operators
- Identity operators
- Membership operators
- Bitwise operators

In this lesson, we will learn the most common ones.

## 1. Arithmetic Operators
Arithmetic operators are used for mathematical calculations.

### Addition
```python
a = 10
b = 5
print(a + b)
```

Output:
```python
15
```

### Subtraction
```python
print(a - b)
```

Output:
```python
5
```

### Multiplication
```python
print(a * b)
```

Output:
```python
50
```

### Division
```python
print(a / b)
```

Output:
```python
2.0
```

### Modulus
Modulus gives the remainder.

```python
print(10 % 3)
```

Output:
```python
1
```

### Exponentiation
```python
print(2 ** 3)
```

Output:
```python
8
```

### Floor Division
Floor division gives the integer quotient.

```python
print(10 // 3)
```

Output:
```python
3
```

## 2. Comparison Operators
Comparison operators compare two values and return True or False.

### Equal to
```python
x = 5
y = 5
print(x == y)
```

Output:
```python
True
```

### Not equal to
```python
print(x != y)
```

Output:
```python
False
```

### Greater than
```python
print(x > 3)
```

Output:
```python
True
```

### Less than
```python
print(x < 3)
```

Output:
```python
False
```

### Greater than or equal to
```python
print(x >= 5)
```

Output:
```python
True
```

### Less than or equal to
```python
print(x <= 5)
```

Output:
```python
True
```

## 3. Assignment Operators
Assignment operators are used to assign values to variables.

```python
a = 10
a += 5
print(a)
```

Output:
```python
15
```

Other examples:
```python
a = 10
a -= 3
print(a)
```

Output:
```python
7
```

```python
a = 10
a *= 2
print(a)
```

Output:
```python
20
```

```python
a = 10
a /= 2
print(a)
```

Output:
```python
5.0
```

## 4. Logical Operators
Logical operators are used to combine conditions.

### AND
```python
x = True
y = False

print(x and y)
```

Output:
```python
False
```

### OR
```python
print(x or y)
```

Output:
```python
True
```

### NOT
```python
print(not x)
```

Output:
```python
False
```

## 5. Identity Operators
Identity operators check whether two variables are the same object.

```python
a = 10
b = 10

print(a is b)
```

Output:
```python
True
```

```python
print(a is not b)
```

Output:
```python
False
```

## 6. Membership Operators
Membership operators test whether a value is in a sequence.

```python
letters = ["a", "b", "c"]

print("a" in letters)
```

Output:
```python
True
```

```python
print("z" not in letters)
```

Output:
```python
True
```

## 7. Bitwise Operators
Bitwise operators work on binary numbers.

```python
a = 5   # 0101
b = 3   # 0011

print(a & b)
print(a | b)
print(a ^ b)
```

Output:
```python
1
7
6
```

## 8. Operator Precedence
Python follows an order of operations, just like mathematics.

Example:
```python
print(10 + 5 * 2)
```

Output:
```python
20
```

Because multiplication is done before addition.

## 9. Example Program Using Operators
```python
a = 15
b = 5

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Modulus:", a % b)

if a > b:
    print("a is greater than b")
else:
    print("a is not greater than b")
```

Output:
```python
Addition: 20
Subtraction: 10
Multiplication: 75
Division: 3.0
Modulus: 0
a is greater than b
```

## 10. Common Beginner Mistakes
### Mistake 1: Using `=` instead of `==`
```python
x = 5
if x = 5:
    print("True")
```
This is wrong because `=` assigns a value, not compares.

Correct:
```python
x = 5
if x == 5:
    print("True")
```

### Mistake 2: Confusing `and` and `or`
```python
a = True
b = False

print(a and b)
print(a or b)
```

Output:
```python
False
True
```
