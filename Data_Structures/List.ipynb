
# Python Lists

A list is a collection of items stored in a single variable.

Lists are used to store multiple values together.

Example:
```python
fruits = ["apple", "banana", "mango"]
print(fruits)
```

Output:
```python
['apple', 'banana', 'mango']
```

## 1. Creating a List
A list is created using square brackets `[]`.

```python
numbers = [10, 20, 30, 40]
print(numbers)
```

Output:
```python
[10, 20, 30, 40]
```

You can also create a list of strings:
```python
names = ["Ali", "Sara", "Hamza"]
print(names)
```

Output:
```python
['Ali', 'Sara', 'Hamza']
```

## 2. Accessing List Elements
Each item in a list has an index number.

Python indexing starts from 0.

```python
fruits = ["apple", "banana", "mango"]
print(fruits[0])
print(fruits[1])
print(fruits[2])
```

Output:
```python
apple
banana
mango
```

## 3. Negative Indexing
Python also supports negative indexing.

```python
fruits = ["apple", "banana", "mango"]
print(fruits[-1])   # last item
print(fruits[-2])   # second last item
```

Output:
```python
mango
banana
```

## 4. Slicing a List
Slicing lets us get a part of the list.

```python
numbers = [1, 2, 3, 4, 5, 6]
print(numbers[1:4])
```

Output:
```python
[2, 3, 4]
```

Another example:
```python
print(numbers[:3])
print(numbers[3:])
```

Output:
```python
[1, 2, 3]
[4, 5, 6]
```

## 5. Changing List Items
Lists are mutable, which means we can change their values.

```python
colors = ["red", "blue", "green"]
colors[1] = "yellow"
print(colors)
```

Output:
```python
['red', 'yellow', 'green']
```

## 6. Adding Elements to a List
We can add items to a list using:
- `append()`
- `insert()`

### append()
```python
fruits = ["apple", "banana"]
fruits.append("mango")
print(fruits)
```

Output:
```python
['apple', 'banana', 'mango']
```

### insert()
```python
fruits.insert(1, "orange")
print(fruits)
```

Output:
```python
['apple', 'orange', 'banana', 'mango']
```

## 7. Removing Elements from a List
We can remove items using:
- `remove()`
- `pop()`
- `del`

### remove()
```python
fruits = ["apple", "banana", "mango"]
fruits.remove("banana")
print(fruits)
```

Output:
```python
['apple', 'mango']
```

### pop()
```python
fruits = ["apple", "banana", "mango"]
fruits.pop(1)
print(fruits)
```

Output:
```python
['apple', 'mango']
```

### del
```python
del fruits[0]
print(fruits)
```

Output:
```python
['mango']
```

## 8. List Length
Use `len()` to check how many items are in a list.

```python
numbers = [10, 20, 30, 40]
print(len(numbers))
```

Output:
```python
4
```

## 9. Looping Through a List
Use a `for` loop to access each item in a list.

```python
fruits = ["apple", "banana", "mango"]

for fruit in fruits:
    print(fruit)
```

Output:
```python
apple
banana
mango
```

## 10. Checking if an Item Exists
Use `in` to check whether a value is present in a list.

```python
fruits = ["apple", "banana", "mango"]

print("apple" in fruits)
```

Output:
```python
True
```

## 11. Sorting a List
Use `sort()` to sort items in ascending order.

```python
numbers = [4, 2, 7, 1]
numbers.sort()
print(numbers)
```

Output:
```python
[1, 2, 4, 7]
```

## 12. Reversing a List
Use `reverse()` to reverse the order.

```python
numbers = [1, 2, 3, 4]
numbers.reverse()
print(numbers)
```

Output:
```python
[4, 3, 2, 1]
```

## 13. Copying a List
Use `copy()` to create a copy of the list.

```python
numbers = [1, 2, 3]
copy_list = numbers.copy()
print(copy_list)
```

Output:
```python
[1, 2, 3]
```

## 14. Concatenating Lists
You can join two lists using `+`.

```python
list1 = [1, 2, 3]
list2 = [4, 5, 6]
result = list1 + list2
print(result)
```

Output:
```python
[1, 2, 3, 4, 5, 6]
```

## 15. List Methods
Some common list methods are:
- `append()`
- `insert()`
- `remove()`
- `pop()`
- `sort()`
- `reverse()`
- `copy()`
- `clear()`

Example:
```python
numbers = [5, 1, 3]
numbers.clear()
print(numbers)
```

Output:
```python
[]
```

## 16. Nested Lists
A list can also contain other lists.

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6]
]

print(matrix[0])
print(matrix[1][2])
```

Output:
```python
[1, 2, 3]
6
```

## 17. List vs String
A list can store many items, while a string stores characters only.

```python
name = "Ali"
letters = ["A", "l", "i"]
print(name)
print(letters)
```

Output:
```python
Ali
['A', 'l', 'i']
```

## 18. Common Mistakes with Lists
### Mistake 1: Using wrong index
```python
fruits = ["apple", "banana"]
print(fruits[2])
```
This will give an IndexError because index 2 does not exist.

### Mistake 2: Forgetting brackets
```python
fruits = apple, banana
```
This is not a list; it is a tuple.

Correct:
```python
fruits = ["apple", "banana"]
```

## 19. Example Program
```python
students = ["Ali", "Sara", "Hamza"]

print("List:", students)
print("First student:", students[0])
print("Length:", len(students))

students.append("Ayesha")
print("After append:", students)
```

Output:
```python
List: ['Ali', 'Sara', 'Hamza']
First student: Ali
Length: 3
After append: ['Ali', 'Sara', 'Hamza', 'Ayesha']
```

## 20. Summary
A list:
- stores multiple values
- uses square brackets `[]`
- allows indexing
- supports sorting, adding, and removing items
- is mutable

Example:
```python
numbers = [10, 20, 30]
numbers.append(40)
print(numbers)
```

Output:
```python
[10, 20, 30, 40]
```
