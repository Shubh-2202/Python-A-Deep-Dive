# Python `filter()` Function

The `filter()` function is used to filter items from an iterable based on a condition.

It is useful when you want to keep only the elements that satisfy a rule.

In this notebook, we will learn:
- what `filter()` does
- syntax
- filtering with lists
- filtering using lambda functions
- practical examples


## 1. What is `filter()`?

`filter()` takes two arguments:
- a function that returns `True` or `False`
- an iterable

It returns only the values for which the condition is `True`.

Example:
```python
numbers = [1, 2, 3, 4, 5, 6]
result = filter(lambda x: x % 2 == 0, numbers)
print(list(result))
```


````python
numbers = [1, 2, 3, 4, 5, 6]
result = filter(lambda x: x % 2 == 0, numbers)
print(list(result))
````

## 2. Syntax

```python
filter(function, iterable)
```

The function should return either `True` or `False`.

Example:
```python
def is_even(x):
    return x % 2 == 0

nums = [1, 2, 3, 4, 5, 6]
print(list(filter(is_even, nums)))
```


````python
def is_even(x):
    return x % 2 == 0

nums = [1, 2, 3, 4, 5, 6]
print(list(filter(is_even, nums)))
````

## 3. Filtering a List

```python
ages = [12, 15, 18, 21, 30, 8]
adults = list(filter(lambda age: age >= 18, ages))
print(adults)
```


````python
ages = [12, 15, 18, 21, 30, 8]
adults = list(filter(lambda age: age >= 18, ages))
print(adults)
````

## 4. `filter()` with Lambda Function

Lambda functions are commonly used with `filter()`.

```python
names = ["Ali", "Sara", "Ayesha", "Zain"]
long_names = list(filter(lambda name: len(name) > 4, names))
print(long_names)
```


````python
names = ["Ali", "Sara", "Ayesha", "Zain", "Bilal"]
long_names = list(filter(lambda name: len(name) > 4, names))
print(long_names)
````

## 5. Filtering Strings

You can also filter strings or other iterables.

```python
words = ["apple", "pear", "banana", "kiwi"]
result = list(filter(lambda word: word.startswith("b"), words))
print(result)
```


````python
words = ["apple", "pear", "banana", "kiwi", "blueberry"]
result = list(filter(lambda word: word.startswith("b"), words))
print(result)
````

## 6. Practical Examples

### Example 1: Keep only positive numbers
```python
numbers = [-3, -1, 0, 4, 7, -9]
positive = list(filter(lambda x: x > 0, numbers))
print(positive)
```

### Example 2: Remove empty strings
```python
items = ["", "apple", "", "banana", "mango"]
clean = list(filter(lambda value: value != "", items))
print(clean)
```


````python
# Example 1: Keep only positive numbers
numbers = [-3, -1, 0, 4, 7, -9]
positive = list(filter(lambda x: x > 0, numbers))
print(positive)

print("---")

# Example 2: Remove empty strings
items = ["", "apple", "", "banana", "mango"]
clean = list(filter(lambda value: value != "", items))
print(clean)
````

## Final Summary

The `filter()` function is used to keep only the items that match a condition.

Key points:
- `filter(function, iterable)`
- returns a filter object
- usually paired with lambda functions
- used to select items based on a condition
- convert to list to view the result

This makes `filter()` very helpful when working with data and conditions.
