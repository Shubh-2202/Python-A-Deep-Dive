# Python Tuples: From Basic to Advanced

A tuple is an ordered, immutable collection of values in Python. It is similar to a list, but once created, it cannot be changed. Tuples are useful when you want fixed data, faster access, and safe storage of values.

In this notebook, we will cover:
- What is a tuple?
- How to create and access tuples
- Tuple operations and methods
- Packing and unpacking
- Advanced usage such as `namedtuple`, nested tuples, and hashing


## 1. What is a Tuple?

A tuple is created using parentheses `()`.

Example:
```python
numbers = (10, 20, 30, 40)
print(numbers)
```

A tuple can hold values of different types:
```python
person = ("Ali", 25, "Lahore")
print(person)
```


````python
# Basic tuple creation
numbers = (10, 20, 30, 40)
print(numbers)

fruits = ("apple", "banana", "mango")
print(fruits)

mixed = (1, "hello", True, 3.14)
print(mixed)

# Single-item tuple needs a trailing comma
single = (42,)
print(single)
````

## 2. Accessing Tuple Elements

Like lists, tuples support indexing. Python indexing starts from `0`.

```python
colors = ("red", "green", "blue")
print(colors[0])
print(colors[1])
print(colors[-1])
```

You can also slice a tuple:
```python
print(colors[1:3])
```


````python
colors = ("red", "green", "blue", "yellow")
print(colors[0])
print(colors[1])
print(colors[-1])
print(colors[1:3])
print(colors[:2])
print(colors[2:])
````

## 3. Why Tuples Are Immutable

Tuples cannot be changed after creation. This makes them safer for fixed data.

```python
nums = (1, 2, 3)
# nums[0] = 99  # This will raise TypeError
```

Tuples are especially useful when you want to prevent accidental modification.


````python
nums = (1, 2, 3)
print(nums)

try:
    nums[0] = 10
except TypeError as e:
    print("TypeError:", e)
````

## 4. Tuple Operations and Methods

Tuples support several useful operations:

- `len(tuple)`
- `tuple + tuple`
- `value in tuple`
- `tuple.count(value)`
- `tuple.index(value)`
- `min(tuple)` / `max(tuple)`

Example:
```python
numbers = (5, 8, 5, 10, 7)
print(len(numbers))
print(numbers.count(5))
print(numbers.index(10))
```


````python
numbers = (5, 8, 5, 10, 7)
print(len(numbers))
print(numbers.count(5))
print(numbers.index(10))
print(numbers + (20, 30))
print(8 in numbers)
print(min(numbers), max(numbers))
````

## 5. Tuple Packing and Unpacking

Tuple packing means creating a tuple without explicitly calling a function. Unpacking means assigning tuple values to variables.

```python
point = (10, 20)
x, y = point
print(x, y)
```

You can also use `*` to collect extra values:
```python
first, *rest = (1, 2, 3, 4, 5)
print(first, rest)
```


````python
# Packing
point = (10, 20)

# Unpacking
x, y = point
print("x =", x, "y =", y)

# Multiple assignment with tuples
a, b, c = (1, 2, 3)
print(a, b, c)

# Collect remaining values
first, *rest = (1, 2, 3, 4, 5)
print(first, rest)
````

## 6. Advanced Tuple Concepts

### Nested Tuples
Tuples can contain other tuples, lists, dictionaries, and more.

### Tuple as a record
Tuples are often used to store fixed records, such as coordinates, RGB values, or database rows.

### Named Tuples
The `collections.namedtuple` class gives tuple elements names, making code easier to read.

### Hashability
Tuples are hashable if all their elements are hashable. This makes them useful as dictionary keys or set elements.


````python
# Nested tuples
matrix = ((1, 2), (3, 4), (5, 6))
print(matrix[1][0])

# Tuple record example
student = ("Ayesha", 21, "Computer Science")
name, age, department = student
print(name, age, department)

# namedtuple
from collections import namedtuple
Point = namedtuple("Point", ["x", "y"])
p = Point(5, 10)
print(p.x, p.y)
print(p[0], p[1])

# Hashability example
coords = {(10, 20), (30, 40)}
print(coords)
````

## 7. When to Use Tuples?

Use tuples when:
- the data should not change
- you want fixed values like coordinates or RGB colors
- you need dictionary keys or set elements
- you want slightly better memory efficiency than lists for some cases

Use lists when:
- you need to modify values often
- you need methods such as `append()`, `remove()`, or `sort()`


## Final Summary

Tuples are immutable, ordered, and efficient. They are best for fixed data and safe value storage. Their use in unpacking, records, and hashing makes them one of the most important built-in data types in Python.

Key ideas:
- Created with `()`
- Indexing and slicing are supported
- Immutable: cannot be changed after creation
- Supports unpacking
- Useful for fixed records and hashable data
- `namedtuple` adds readability
