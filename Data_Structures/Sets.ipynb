# Python Sets for Beginners

A set is a collection of unique items in Python.

Sets are used to store data without duplicates. They are very useful when you want to keep only unique values.

Example:
```python
numbers = {1, 2, 3, 4, 4, 5}
print(numbers)
```

Output:
```python
{1, 2, 3, 4, 5}
```

Notice that `4` appears only once because sets do not allow duplicate values.


## 1. What is a Set?

A set is created using curly braces `{}`.

```python
fruits = {"apple", "banana", "mango"}
print(fruits)
```

A set can contain strings, numbers, and other values, but each value must be unique.


````python
# Creating a set
numbers = {10, 20, 30, 40}
print(numbers)

# Duplicate values are removed automatically
numbers_with_duplicates = {1, 2, 2, 3, 3, 4}
print(numbers_with_duplicates)

# Set of strings
colors = {"red", "blue", "green"}
print(colors)
````

## 2. Accessing Set Items

Sets are unordered, which means the items do not have a fixed position.

Because of this, you cannot access items by index like you can with a list or tuple.

```python
fruits = {"apple", "banana", "mango"}
# fruits[0]  # This will raise an error
```

To check if an item exists in a set, use the `in` keyword.
```python
print("apple" in fruits)
```


````python
fruits = {"apple", "banana", "mango"}
print("apple" in fruits)
print("grape" in fruits)

# You cannot access by index
try:
    print(fruits[0])
except TypeError as e:
    print("TypeError:", e)
````

## 3. Common Set Operations

Sets support many useful operations.

### add()
Adds one item to the set.

```python
numbers = {1, 2, 3}
numbers.add(4)
print(numbers)
```

### remove()
Removes an item from the set.

```python
numbers.remove(2)
print(numbers)
```

### discard()
Also removes an item, but it does not raise an error if the item is not found.

### pop()
Removes and returns a random item from the set.


````python
numbers = {1, 2, 3}
numbers.add(4)
print("After add:", numbers)

numbers.remove(2)
print("After remove:", numbers)

numbers.discard(99)
print("After discard:", numbers)

popped = numbers.pop()
print("Popped item:", popped)
print("Remaining set:", numbers)
````

## 4. Set Operations

Python sets support mathematical set operations such as union, intersection, and difference.

### Union `|`
Combines all unique values from both sets.

```python
A = {1, 2, 3}
B = {3, 4, 5}
print(A | B)
```

### Intersection `&`
Keeps only the values that are common in both sets.

```python
print(A & B)
```

### Difference `-`
Shows items in the first set that are not in the second set.


````python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print("Union:", A | B)
print("Intersection:", A & B)
print("Difference A-B:", A - B)
print("Difference B-A:", B - A)
````

## 5. Looping Through a Set

You can loop through a set using a `for` loop.

```python
fruits = {"apple", "banana", "mango"}
for fruit in fruits:
    print(fruit)
```


````python
fruits = {"apple", "banana", "mango"}
for fruit in fruits:
    print(fruit)
````

## 6. Why Use Sets?

Sets are useful when you need:
- unique values only
- fast membership checks (`in`)
- mathematical operations like union and intersection
- removing duplicates from data

Example:
```python
data = [1, 2, 2, 3, 4, 4, 5]
unique_data = set(data)
print(unique_data)
```


````python
data = [1, 2, 2, 3, 4, 4, 5, 5]
unique_data = set(data)
print("Original list:", data)
print("Unique values:", unique_data)
````

## Final Summary

A set is a collection of unique values. It is unordered, does not allow duplicates, and is great for membership testing and mathematical set operations.

Key points:
- created with `{}`
- no duplicate values
- unordered
- supports `add()`, `remove()`, `discard()`, `pop()`
- supports union, intersection, and difference
- very useful for filtering repeated data
