# Python `map()` Function

The `map()` function is used to apply a function to every item in an iterable such as a list, tuple, or set.

It is one of the most useful built-in functions in Python for transforming data.

In this notebook, we will learn:
- what `map()` does
- basic syntax
- map with lists
- map with lambda functions
- map with tuples and strings
- practical examples


## 1. What is `map()`?

`map()` takes two arguments:
- a function
- an iterable

It applies the function to each item and returns a map object.

Example:
```python
numbers = [1, 2, 3, 4]
result = map(lambda x: x * 2, numbers)
print(list(result))
```


````python
numbers = [1, 2, 3, 4]
result = map(lambda x: x * 2, numbers)
print(list(result))
````

## 2. Syntax

```python
map(function, iterable)
```

The `function` is applied to every item in the iterable.

Example:
```python
def square(x):
    return x * x

nums = [1, 2, 3, 4]
print(list(map(square, nums)))
```


````python
def square(x):
    return x * x

nums = [1, 2, 3, 4]
print(list(map(square, nums)))
````

## 3. Using `map()` with a List

```python
numbers = [5, 10, 15, 20]
result = map(lambda x: x + 5, numbers)
print(list(result))
```


````python
numbers = [5, 10, 15, 20]
result = map(lambda x: x + 5, numbers)
print(list(result))
````

## 4. `map()` with Lambda Function

Lambda functions work very well with `map()`.

```python
names = ["ali", "sara", "ahmed"]
upper_names = list(map(lambda name: name.upper(), names))
print(upper_names)
```


````python
names = ["ali", "sara", "ahmed"]
upper_names = list(map(lambda name: name.upper(), names))
print(upper_names)
````

## 5. `map()` with Multiple Iterables

`map()` can also take more than one iterable.

```python
a = [1, 2, 3]
b = [10, 20, 30]
result = list(map(lambda x, y: x + y, a, b))
print(result)
```


````python
a = [1, 2, 3]
b = [10, 20, 30]
result = list(map(lambda x, y: x + y, a, b))
print(result)
````

## 6. `map()` with Tuples and Strings

`map()` works with any iterable, including tuples and strings.

```python
nums = (1, 2, 3, 4)
print(list(map(lambda x: x * 3, nums)))

text = "python"
print(list(map(lambda ch: ch.upper(), text)))
```


````python
nums = (1, 2, 3, 4)
print(list(map(lambda x: x * 3, nums)))

text = "python"
print(list(map(lambda ch: ch.upper(), text)))
````

## 7. Practical Examples

### Example 1: Convert temperatures
```python
celsius = [0, 10, 20, 30]
fahrenheit = list(map(lambda c: (9 / 5) * c + 32, celsius))
print(fahrenheit)
```

### Example 2: Square each number
```python
numbers = [2, 4, 6, 8]
squares = list(map(lambda x: x ** 2, numbers))
print(squares)
```


````python
# Example 1: Convert temperatures
celsius = [0, 10, 20, 30]
fahrenheit = list(map(lambda c: (9 / 5) * c + 32, celsius))
print(fahrenheit)

print("---")

# Example 2: Square each number
numbers = [2, 4, 6, 8]
squares = list(map(lambda x: x ** 2, numbers))
print(squares)
````

## Final Summary

The `map()` function is a powerful way to transform every item in an iterable using a function.

Key points:
- `map(function, iterable)`
- returns a map object
- often used with lambda functions
- works with lists, tuples, and strings
- convert to list to see the output

This makes `map()` very useful in data processing and functional programming.
