# Python Dictionaries: From Basic to Intermediate

A dictionary is a collection of key-value pairs. It is one of the most useful data types in Python because it stores data in a structured way.

In this notebook, we will learn:
- what a dictionary is
- how to create dictionaries
- access values and keys
- add and update items
- remove values
- loop through dictionaries
- common intermediate examples


## 1. What is a Dictionary?

A dictionary stores data in pairs of `key: value`.

Example:
```python
student = {"name": "Ali", "age": 20, "city": "Karachi"}
print(student)
```

Here:
- `name` is the key
- `Ali` is the value


````python
student = {"name": "Ali", "age": 20, "city": "Karachi"}
print(student)

person = {"name": "Sara", "country": "Pakistan"}
print(person)
````

## 2. Accessing Dictionary Values

You can access the value of a key by using square brackets `[]`.

```python
student = {"name": "Ali", "age": 20}
print(student["name"])
print(student["age"])
```

You can also use the `get()` method.
```python
print(student.get("name"))
```


````python
student = {"name": "Ali", "age": 20, "city": "Karachi"}
print(student["name"])
print(student["age"])
print(student.get("city"))
print(student.get("country", "Not found"))
````

## 3. Keys and Values

A dictionary has keys and values. You can get them separately.

```python
student = {"name": "Ali", "age": 20, "city": "Karachi"}
print(student.keys())
print(student.values())
print(student.items())
```


````python
student = {"name": "Ali", "age": 20, "city": "Karachi"}
print(student.keys())
print(student.values())
print(student.items())
````

## 4. Adding and Updating Values

Use assignment to add or update data in a dictionary.

```python
student = {"name": "Ali"}
student["age"] = 20
student["name"] = "Ahmed"
print(student)
```


````python
student = {"name": "Ali"}
student["age"] = 20
student["city"] = "Karachi"
student["name"] = "Ahmed"
print(student)
````

## 5. Removing Items

We can remove items from a dictionary using methods like `pop()`, `popitem()`, and `clear()`.

```python
student = {"name": "Ali", "age": 20}
student.pop("age")
print(student)
```


````python
student = {"name": "Ali", "age": 20, "city": "Karachi"}
student.pop("age")
print(student)

student.popitem()
print(student)

student.clear()
print(student)
````

## 6. Looping Through a Dictionary

You can loop through keys, values, or items.

```python
student = {"name": "Ali", "age": 20, "city": "Karachi"}

for key in student:
    print(key)

for key, value in student.items():
    print(key, value)
```


````python
student = {"name": "Ali", "age": 20, "city": "Karachi"}

for key in student:
    print("Key:", key)

print("---")
for key, value in student.items():
    print(key, ":", value)
````

## 7. Intermediate Examples

### Example 1: Count word frequencies
```python
text = "apple banana apple orange banana apple"
words = text.split()
counts = {}

for word in words:
    counts[word] = counts.get(word, 0) + 1

print(counts)
```

### Example 2: Merge dictionaries
```python
a = {"x": 1, "y": 2}
b = {"y": 3, "z": 4}
a.update(b)
print(a)
```

### Example 3: Nested dictionary
```python
students = {
    "Ali": {"age": 20, "city": "Karachi"},
    "Sara": {"age": 22, "city": "Lahore"}
}
print(students["Ali"]["city"])
```


````python
# Example 1: Count word frequencies
text = "apple banana apple orange banana apple"
words = text.split()
counts = {}

for word in words:
    counts[word] = counts.get(word, 0) + 1

print(counts)

print("---")

# Example 2: Merge dictionaries
a = {"x": 1, "y": 2}
b = {"y": 3, "z": 4}
a.update(b)
print(a)

print("---")

# Example 3: Nested dictionary
students = {
    "Ali": {"age": 20, "city": "Karachi"},
    "Sara": {"age": 22, "city": "Lahore"}
}
print(students["Ali"]["city"])
````

## Final Summary

Dictionaries are powerful because they store data as key-value pairs. They are used widely in Python for mapping, counting, grouping, and organizing data.

Key points:
- created with `{}`
- keys must be unique
- values can be numbers, strings, lists, or even other dictionaries
- use `[]` or `get()` to access values
- use `update()`, `pop()`, and `clear()` to manage data
- loop with `items()`, `keys()`, and `values()`

Dictionaries are essential for beginner and intermediate Python programming.
