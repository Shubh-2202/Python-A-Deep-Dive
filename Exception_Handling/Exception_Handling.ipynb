# Python Exception Handling: Beginner to Intermediate

An exception is an error that happens while a program is running. Exception handling lets us respond to expected errors so a program can show a useful message or continue safely.

In this notebook, you will learn about `try`, `except`, `else`, `finally`, raising exceptions, and creating a custom exception.

## 1. What Is an Exception?

Python raises an exception when it cannot complete an operation. For example, dividing by zero raises `ZeroDivisionError`, and converting non-numeric text to an integer raises `ValueError`.

If an exception is not handled, Python stops the current program flow and displays an error message called a traceback.

## 2. The `try` and `except` Blocks

Put code that might fail inside a `try` block. Use an `except` block to handle a specific exception.

```python
try:
    number = int("not a number")
except ValueError:
    print("Please provide digits that form a whole number.")
```

Prefer catching a specific exception, such as `ValueError`, instead of catching every possible error.

````python
text = "hello"

try:
    number = int(text)
    print("Number:", number)
except ValueError:
    print("That text cannot be converted to an integer.")
````

## 3. Handling More Than One Exception

Use separate `except` blocks when different errors need different responses. You can also handle several exception types together when the response should be the same.

````python
values = [10, 0]

try:
    position = int("1")
    answer = values[0] / values[position]
    print(answer)
except ValueError:
    print("The position must be an integer.")
except ZeroDivisionError:
    print("The selected value cannot be zero.")
except IndexError:
    print("That position is outside the list.")
````

## 4. The `else` and `finally` Blocks

- `else` runs only if the `try` block finishes without an exception.
- `finally` runs whether an exception happened or not. It is commonly used for cleanup. For files, `with open(...)` is usually the simpler and safer cleanup pattern.

````python
text = "42"

try:
    number = int(text)
except ValueError:
    print("Please enter a whole number.")
else:
    print("Conversion succeeded:", number)
finally:
    print("Conversion attempt finished.")
````

## 5. Reading the Exception Message

You can bind an exception to a variable with `as`. This can help when you need to show or record a useful error message. Avoid exposing sensitive internal details to end users in real applications.

````python
try:
    result = 10 / 0
except ZeroDivisionError as error:
    print("Could not divide:", error)
````

## 6. Raising an Exception

Use `raise` when your code detects invalid data or a situation it cannot safely handle. Choose an exception type that describes the problem.

````python
def set_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative.")
    return age

try:
    print(set_age(-3))
except ValueError as error:
    print("Invalid age:", error)
````

## 7. Creating a Custom Exception

For a specific problem in your program, you can create a custom exception by inheriting from `Exception`. Custom exceptions can make errors in larger programs easier to understand.

````python
class InsufficientFundsError(Exception):
    """Raised when a withdrawal is greater than the available balance."""


def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientFundsError("Not enough money in the account.")
    return balance - amount


try:
    new_balance = withdraw(50, 75)
except InsufficientFundsError as error:
    print("Withdrawal failed:", error)
````

## 8. Good Exception-Handling Practices

- Catch only exceptions you expect and know how to handle.
- Keep the `try` block small so you know which operation failed.
- Do not use a bare `except:` or silently ignore an error with `pass`.
- Use `finally` or context managers such as `with` when cleanup is needed.
- Give clear, helpful messages and preserve the original error when re-raising.
- Do not use exceptions to replace ordinary checks when a simple condition is clearer.

````python
def divide_text(first_text, second_text):
    try:
        first_number = float(first_text)
        second_number = float(second_text)
    except ValueError:
        print("Both inputs must be numbers.")
        return None

    if second_number == 0:
        print("The second number must not be zero.")
        return None

    return first_number / second_number


print(divide_text("12", "3"))
print(divide_text("hello", "3"))
print(divide_text("12", "0"))
````

## Final Summary

- Put risky operations in a `try` block.
- Handle expected errors with specific `except` blocks.
- Use `else` for code that should run after success.
- Use `finally` for code that must run either way.
- Use `raise` to report invalid or unsupported situations.
- Create custom exception classes for meaningful application-specific errors.

Exception handling helps programs respond clearly and safely when something goes wrong.
